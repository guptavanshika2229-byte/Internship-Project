I'm picking this lane because Notebooks 1 and 2 already showed two things worth
building on: a hand-written "stale x visible" rule can be sharp at the very top of
a ranked list (Precision@20 = 0.900), but it only fires on a tiny sliver of pages —
just 17 of 30,000 (0.1%) meet both the staleness and visibility thresholds at once.

model, by contrast, can rank the other 99.9% of pages using softer, combined signals
instead of one hard rule. This lane lets me test whether a model-based ranking
finds good candidates beyond what the hand rule can see, while staying anchored to
something I can explain page by page.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision this improves: which page a content editor should review first this
week, out of thousands, given limited review capacity.

Who acts, and how: a content editor or SEO strategist works down a ranked
queue and decides, per page, to refresh, expand, protect, or leave it alone.

Cost of a wrong call: false positive — editor spends an hour reviewing a page
that wasn't actually a priority, wasted time but low risk. False negative — a
genuinely declining, high-traffic page gets missed and keeps losing visibility
for another review cycle, which is the more expensive error since traffic is
harder to win back than editor time is to spend. This asymmetry is why
Precision@K, not raw accuracy, is the right metric here.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape)

stale = df["days_since_last_update"] >= 180
visible = df["impressions_90d"] >= 500
both = stale & visible
print(f"Pages matching hand rule (stale AND visible): {both.sum()} of {len(df)} ({both.mean()*100:.1f}%)")

declining = df["trend_direction"].str.lower().eq("down")
print(f"Declining rate among hand-rule matches: {declining[both].mean():.3f}")
print(f"Declining rate overall: {declining.mean():.3f}")
```

position, CTR) and a page later showing a declining trend bucket, on this
